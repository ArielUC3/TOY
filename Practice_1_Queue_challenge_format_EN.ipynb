# 🧱 Base code: `Queue` class

The following class implements a FIFO queue using a Python list.

**Run this cell before starting the challenges and do not modify it.**

````python
class Queue:
    """Implementation of a FIFO queue using a native Python list."""

    def __init__(self):
        """Initializes an empty queue."""
        self.items = []

    def __len__(self):
        """Returns the number of elements in the queue."""
        return len(self.items)

    def isEmpty(self):
        """Returns True if the queue is empty, False otherwise."""
        return len(self) == 0

    def __str__(self):
        """Returns a string representation of the queue elements."""
        return str(self.items)

    def enqueue(self, e):
        """Adds element e to the rear (tail) of the queue."""
        self.items.append(e)

    def dequeue(self):
        """Removes and returns the first element (head) of the queue."""
        if self.isEmpty():
            print("Error: The queue is empty")
            return None
        return self.items.pop(0)

    def front(self):
        """Returns the first element without removing it."""
        if self.isEmpty():
            print("Error: The queue is empty")
            return None
        return self.items[0]
````

# 🏆 Challenge 1. Inspecting the last element without modifying the queue

🎯 **Goal:** Fix an implementation that accidentally destroys the queue contents.

🕒 **Estimated time:** 10 minutes.

🟢 **Difficulty:** Easy.

## 📝 Problem statement

A student wrote `get_last_incorrect()` to inspect the **last enqueued element** —the element at the rear or *tail* of the queue.

The code has two problems:

1. It repeatedly calls `dequeue()` and therefore **empties the original queue**.
2. It does not explicitly handle an empty queue.

Implement `get_last(queue)` so that:

- It returns `None` if the queue is empty.
- It returns the last element if the queue contains elements.
- After the function finishes, the queue contains **exactly the same elements in the same order**.

> **💡 Hint**
>
> You can traverse a queue without losing its elements if every dequeued element is enqueued again at the rear.

### 🔎 Incorrect code to analyze

````python
def get_last_incorrect(queue: Queue):
    last = None

    while not queue.isEmpty():
        last = queue.dequeue()

    return last
````

### 💻 Your solution

````python
def get_last(queue: Queue):
    if len(queue) == 0:
        return None

    n = len(queue)

    for i in range(n - 1):
        element = queue.dequeue()
        queue.enqueue(element)

    last = queue.dequeue()
    queue.enqueue(last)

    return last
````

### ✅ Automatic checking

````python
q = Queue()
q.enqueue(10)
q.enqueue(20)
q.enqueue(30)

initial_state = str(q)
last = get_last(q)

assert last == 30, f"Expected 30, but got {last}"
assert len(q) == 3, "The queue must still contain three elements"
assert q.front() == 10, "The front element must still be 10"
assert str(q) == initial_state, "The queue contents and order must not change"

empty = Queue()
assert get_last(empty) is None
assert empty.isEmpty()

print("Tests passed")
````

# 🏆 Challenge 2. Round-Robin CPU scheduler

🎯 **Goal:** Simulate a CPU scheduler using a FIFO queue.

🕒 **Estimated time:** 20 minutes.

🟡 **Difficulty:** Intermediate.

## 📝 Problem statement

An operating system uses the **Round-Robin** algorithm to allocate CPU time to processes.

Each process is stored as a tuple:

```python
(process_name, remaining_execution_time)
```

The scheduler uses a fixed `quantum` and works as follows:

1. Dequeue the first available process.
2. If its remaining time is **greater than the quantum**, subtract the quantum and enqueue the process again at the rear.
3. If its remaining time is **less than or equal to the quantum**, the process finishes and is not re-enqueued.
4. Continue until there are no pending processes.

Complete `run_simulation(self)` so that it returns the process names **in the exact order in which they finish**.

> **💡 Hint**
>
> In each iteration, you only need to process the element at the front of the queue.

### 💻 Your solution

````python
class CPUScheduler:
    def __init__(self, quantum: int = 2):
        self.process_queue = Queue()
        self.quantum = quantum

    def add_process(self, name: str, time: int):
        """Adds a new process to the execution queue."""
        self.process_queue.enqueue((name, time))

    def run_simulation(self) -> list:
        """
        Runs all processes using Round-Robin and returns their names
        in the exact order in which they finish.
        """
        finished_processes = []

        while len(self.process_queue) > 0:
          a = self.process_queue.dequeue()
          if a[1] <= self.quantum:
            finished_processes.append(a[0])

          else:
            a = (a[0], a[1] - self.quantum)
            self.process_queue.enqueue((a))

        return finished_processes
````

### ✅ Automatic checking

````python
scheduler = CPUScheduler(quantum=2)
scheduler.add_process("P1", 5)
scheduler.add_process("P2", 2)
scheduler.add_process("P3", 1)

order = scheduler.run_simulation()

assert order == ["P2", "P3", "P1"], (
    f"Expected ['P2', 'P3', 'P1'], but got {order}"
)
assert scheduler.process_queue.isEmpty(), (
    "The process queue must be empty when the simulation finishes"
)

print("Tests passed")
````

# 🏆 Challenge 3. Extending the `Queue` class

🎯 **Goal:** Create a subclass with additional queue operations.

🕒 **Estimated time:** 15 minutes.

🟡 **Difficulty:** Intermediate.

## 📝 Problem statement

Create a subclass called `AdvancedQueue` that inherits from `Queue` and implements:

1. `rear()`
   - Returns the last enqueued element.
   - Does not remove it.
   - Returns `None` if the queue is empty.

2. `rotate(k)`
   - Performs `k` rotations.
   - Each rotation moves the front element to the rear.

3. `multi_dequeue(k)`
   - Dequeues up to `k` elements.
   - Returns the removed elements in a list, in dequeue order.
   - If `k` is greater than the queue size, removes all available elements.

> **💡 Hint**
>
> You can implement `rotate()` by reusing only `dequeue()` and `enqueue()`.

### 💻 Your solution

````python
class AdvancedQueue(Queue):
    """Queue subclass with additional methods."""

    def rear(self):
        """Returns the element at the rear without removing it."""
        if len(self.items) == 0:
          return None
        else:
          for i in range(len(self.items) -1):
            self.enqueue(self.dequeue())
          a = self.dequeue()
          self.enqueue(a)
          return a


    def rotate(self, k: int):
        """Rotates the queue k times by moving the front to the rear."""
        for i in range(k):
          self.enqueue(self.dequeue())

        pass

    def multi_dequeue(self, k: int) -> list:
        """Dequeues and returns up to k elements."""
        errased = []
        if k > len(self.items):
          for i in range(len(self.items)):
             a = self.dequeue()
             errased.append(a)

        else:
          for i in range(k):
            a = self.dequeue()
            errased.append(a)

        return errased
````

### ✅ Automatic checking

````python
aq = AdvancedQueue()

for x in [1, 2, 3, 4]:
    aq.enqueue(x)

assert aq.rear() == 4
assert len(aq) == 4, "rear() must not modify the queue"

aq.rotate(1)
assert str(aq) == "[2, 3, 4, 1]"
assert aq.front() == 2

removed = aq.multi_dequeue(2)
assert removed == [2, 3]
assert str(aq) == "[4, 1]"
assert len(aq) == 2

# If k is greater than the size, remove all available elements.
assert aq.multi_dequeue(10) == [4, 1]
assert aq.isEmpty()
assert aq.rear() is None

print("Tests passed")
````

# 🏆 Challenge 4. Hot Potato game

🎯 **Goal:** Apply the Queue ADT to simulate a circular elimination process.

🕒 **Estimated time:** 20 minutes.

🟡 **Difficulty:** Intermediate.

## 📝 Problem statement

In the **Hot Potato** game, a group of players forms a circle and passes an object around.

The simulation must work as follows:

1. Enqueue all players in the given order.
2. Perform `num_passes` transfers. Each transfer moves the front player to the rear of the queue.
3. After the transfers, eliminate the player at the front.
4. Repeat until only one player remains.
5. Return the winner's name.

Implement:

```python
play_hot_potato(players, num_passes)
```

using an instance of `Queue`.

> **💡 Hint**
>
> Circular movement can be simulated by dequeuing the front player and immediately enqueueing that player at the rear.

### 💻 Your solution

````python
def play_hot_potato(players: list, num_passes: int) -> str:
    queue = Queue()
    for n in players:
      queue.enqueue(n)

    while len(queue) > 1:
      for i in range(num_passes):
        queue.enqueue(queue.dequeue())
      a = queue.dequeue()
    return queue.dequeue()
    pass
````

### ✅ Automatic checking

````python
players = ["Ana", "Carlos", "Diana", "Esteban"]

winner = play_hot_potato(players, num_passes=3)

assert winner == "Carlos", (
    f"Expected 'Carlos', but got '{winner}'"
)

# Minimum case.
assert play_hot_potato(["Ana"], num_passes=5) == "Ana"

print("Tests passed")
````
