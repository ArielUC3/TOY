# 🏆 Challenge 2. Summary of a List of Numbers

🎯 **Objective:** Implement `sign_summary(values)` to return a dictionary with the number of positive, negative, and zero values.

Example:

```python
sign_summary([3, -1, 0, 7, -4])
# {"positives": 2, "negatives": 2, "zeros": 1}
```

🕒 **Estimated time:** 8 minutes

🟢 **Difficulty:** Easy

### 🧠 Before Designing
- What counters do you need?
- What should happen if the list is empty?
- How do you classify each value?


````python
def sign_summary(values):
    dict = {
        "positives": 0,
        "negatives": 0,
        "zeros": 0
    }
    pcount = 0
    ncount = 0
    zcount = 0
    for value in values:
      if value > 0:
        pcount = pcount + 1
      elif value < 0:
        ncount = ncount + 1
      else:
        zcount = zcount + 1

    dict["positives"] = pcount
    dict["negatives"] = ncount
    dict["zeros"] = zcount


    return dict
````

Run the following cell to check your solution.


````python
assert sign_summary([3, -1, 0, 7, -4]) == {
    "positives": 2,
    "negatives": 2,
    "zeros": 1,
}
assert sign_summary([]) == {
    "positives": 0,
    "negatives": 0,
    "zeros": 0,
}
assert sign_summary([0, 0, 0]) == {
    "positives": 0,
    "negatives": 0,
    "zeros": 3,
}
assert sign_summary([-5, 2, 8]) == {
    "positives": 2,
    "negatives": 1,
    "zeros": 0,
}
print("Tests passed")
````

# 🏆 Challenge 3. Cumulative Totals

🎯 **Objective:** Implement `cumulative_totals(values)` to return a new list where each position contains the sum of all elements up to that point.

Example:


🟢 **Difficulty:** Easy

### 🧠 Before Designing
- What variable can store the accumulated sum?
- Should the original list be modified?

````python
def cumulative_totals(values):
    flist = []
    sum = 0
    for value in values:
      sum = sum + value
      flist.append(sum)
    return flist
````

# 🏆 Challenge 5. Compressing Consecutive Values

🎯 **Objective:** Implement `compress_consecutive(values)` to remove consecutive repetitions, keeping a single copy of each block.

Example:

```python
compress_consecutive([1, 1, 1, 2, 2, 3, 1, 1])
# [1, 2, 3, 1]
```

🕒 **Estimated time:** 10 minutes

🟡 **Difficulty:** Medium

### 🧠 Before Designing
- What is the difference between any duplicate and a consecutive duplicate?
- What value should each element be compared with?
- What happens if the list is empty?


````python
def compress_consecutive(values):
    f_list = []

    if len(values) == 0:
        return f_list

    f_list.append(values[0])

    for i in range(len(values) - 1):
        if values[i+1] != values[i]:
            f_list.append(values[i+1])

    return f_list
````

````python
assert compress_consecutive([1, 1, 1, 2, 2, 3, 1, 1]) == [1, 2, 3, 1]
assert compress_consecutive([]) == []
assert compress_consecutive([4]) == [4]
assert compress_consecutive(["a", "a", "b", "b", "a"]) == ["a", "b", "a"]
print("Tests passed")
````

# 🏆 Challenge 6. Lengths of Active Runs

🎯 **Objective:** Implement `active_runs(states)` to return the lengths of all consecutive blocks of `True` values.

Example:

```python
active_runs([True, True, False, True, False, True, True, True])
# [2, 1, 3]
```

🕒 **Estimated time:** 12 minutes

🟡 **Difficulty:** Medium

### 🧠 Before Designing
- How do you count the length of a current run?
- When should that run be added to the result?
- What happens if the list ends with `True` values?


````python
def active_runs(states):
    f_listt = []

    if len(states) == 0:
        return f_listt

    length = 0

    for i in range(len(states)):
        if states[i] == True:
            length = length + 1
        else:
            if length > 0:
                f_listt.append(length)
            length = 0

    if length > 0:
        f_listt.append(length)

    return f_listt
````

Run the following cell to check your solution.


````python
assert active_runs(
    [True, True, False, True, False, True, True, True]
) == [2, 1, 3]
assert active_runs([]) == []
assert active_runs([False, False]) == []
assert active_runs([True, True, True]) == [3]
assert active_runs([False, True, False, True]) == [1, 1]
print("Tests passed")
````

- What happens with the last block if it is shorter?

````python
def split_into_blocks(values, size):
    blocks = []

    for i in range(0, len(values), size):
        blocks.append(values[i:i + size])

    return blocks
````

# 🏆 Challenge 8. Rotating a List

🎯 **Objective:** Implement `rotate_right(values, k)` to return a new list rotated `k` positions to the right.

Example:

```python
rotate_right([1, 2, 3, 4, 5], 2)
# [4, 5, 1, 2, 3]
```

The function must support values of `k` greater than the length of the list.

🕒 **Estimated time:** 12 minutes

🟡 **Difficulty:** Medium

### 🧠 Before Designing
- What does rotating by the length of the list mean?
- How can you reduce `k` using the modulo operator `%`?
- How can you combine slices to build the result?


````python
def rotate_right(values, k):
    if len(values) == 0:
        return []

    k = k % len(values)

    return values[-k:] + values[:-k]
````

Run the following cell to check your solution.


````python
data = [1, 2, 3, 4, 5]
assert rotate_right(data, 2) == [4, 5, 1, 2, 3]
assert data == [1, 2, 3, 4, 5]
assert rotate_right([], 4) == []
assert rotate_right([1, 2, 3], 0) == [1, 2, 3]
assert rotate_right([1, 2, 3], 4) == [3, 1, 2]
print("Tests passed")
````

# 🏆 Challenge 9. Interleaving Two Lists

🎯 **Objective:** Implement `interleave(a, b)` to alternate elements from both lists.

When one list is exhausted, append the remaining elements from the other list at the end.

Example:

```python
interleave([1, 2, 3], ["a", "b"])
# [1, "a", 2, "b", 3]
```

🕒 **Estimated time:** 12 minutes

🟡 **Difficulty:** Medium

### 🧠 Before Designing
- Which list is longer?
- How can you avoid accessing a position that does not exist?
- What order should the elements be added in?


````python
def interleave(a, b):
    lst = []

    i = 0

    while i < len(a) or i < len(b):
        if i < len(a):
            lst.append(a[i])

        if i < len(b):
            lst.append(b[i])

        i = i + 1

    return lst
````

Run the following cell to check your solution.


````python
assert interleave([1, 2, 3], ["a", "b"]) == [1, "a", 2, "b", 3]
assert interleave([], [1, 2]) == [1, 2]
assert interleave([1, 2], []) == [1, 2]
assert interleave([1], ["a", "b", "c"]) == [1, "a", "b", "c"]
assert interleave([], []) == []
print("Tests passed")
````

# 🏆 Challenge 11. Validating a Rectangular Matrix

🎯 **Objective:** Implement `is_rectangular(matrix)` to indicate whether all rows have the same length.

Consider an empty matrix to be rectangular.

Examples:

```python
is_rectangular([[1, 2], [3, 4]])  # True
is_rectangular([[1, 2], [3]])     # False
```

🕒 **Estimated time:** 10 minutes

🟡 **Difficulty:** Medium

### 🧠 Before Designing
- What length should be used as a reference?
- What special case appears when the matrix is empty?
- When can the algorithm stop early?


````python
def is_rectangular(matrix):
    if len(matrix) == 0:
        return True

    for i in range(len(matrix) - 1):
        if len(matrix[i]) != len(matrix[i + 1]):
            return False

    return True
````

Run the following cell to check your solution.


````python
assert is_rectangular([]) is True
assert is_rectangular([[]]) is True
assert is_rectangular([[1, 2], [3, 4]]) is True
assert is_rectangular([[1, 2], [3]]) is False
assert is_rectangular([[1], [2], [3]]) is True
print("Tests passed")
````

# 🏆 Challenge 12. Calculating the Total of a Purchase

🎯 **Objective:** Implement `purchase_total(products)`.

Each product is a dictionary with:

- `price`
- `quantity`
- `discount`, expressed as a percentage between 0 and 100

The function must return the final total rounded to 2 decimal places.

Example:

```python
purchase_total([
    {"price": 10, "quantity": 2, "discount": 0},
    {"price": 5, "quantity": 3, "discount": 20},
])
# 32.0
```

🕒 **Estimated time:** 12 minutes

🟡 **Difficulty:** Medium

### 🧠 Before Designing
- How do you calculate the subtotal for each product?
- How do you apply a percentage discount?
- How do you accumulate the final total?


````python
def purchase_total(products):
  total = 0
  for product in products:
    total1 = (product["price"] * product["quantity"])
    total2 = total1 - (total1  * (product["discount"] / 100 ))
    total += total2

  return round(total, 2)
````

Run the following cell to check your solution.


````python
assert purchase_total([]) == 0.0
assert purchase_total([
    {"price": 10, "quantity": 2, "discount": 0},
    {"price": 5, "quantity": 3, "discount": 20},
]) == 32.0
assert purchase_total([
    {"price": 19.99, "quantity": 1, "discount": 10},
]) == 17.99
assert purchase_total([
    {"price": 3.5, "quantity": 4, "discount": 50},
]) == 7.0
print("Tests passed")
````

# 🏆 Challenge 13. Validating an ISBN-10

🎯 **Objective:** Implement `valid_isbn10(isbn)` to check an ISBN-10 written as a string of exactly 10 digits.

An ISBN-10 is valid when:

```text
1?d1 + 2?d2 + 3?d3 + ... + 10?d10
```

is a multiple of 11.

In this challenge, ISBNs ending in `X` are not considered.

🕒 **Estimated time:** 15 minutes

🔴 **Difficulty:** Difficult

### 🧠 Before Designing
- How do you check that the string has exactly 10 characters?
- How do you verify that all characters are digits?
- How do you calculate the weighted sum?


````python
def valid_isbn10(isbn):
    if len(isbn) != 10:
        return False

    for digit in isbn:
        if not digit.isdigit():
            return False

    total = 0

    for i in range(10):
        total += (i + 1) * int(isbn[i])

    return total % 11 == 0
````

Run the following cell to check your solution.


````python
assert valid_isbn10("0306406152") is True
assert valid_isbn10("0131103628") is True
assert valid_isbn10("0306406153") is False
assert valid_isbn10("") is False
assert valid_isbn10("123") is False
assert valid_isbn10("03064A6152") is False
print("Tests passed")
````
